```python
import requests
import json
```

```python
ROBOT_IP = "10.64.7.132"
PROTOCOL_FILE = "scripts_library/Adjusted_Tip.py"

HEADERS = {"opentrons-version": "3"}

protocols_url = f"http://{ROBOT_IP}:31950/protocols"
protocol_file_payload = open(PROTOCOL_FILE, "rb")

r = requests.post(
	url=protocols_url,
	headers=HEADERS,
	files={"files": protocol_file_payload}
	)

r_dict = json.loads(r.text)
protocol_id = r_dict["data"]["id"]
print(f"Protocol ID:\n{protocol_id}")

protocol_file_payload.close()
```

```python
runs_url = f"http://{ROBOT_IP}:31950/runs"
protocol_id_payload = json.dumps({"data":{"protocolId": protocol_id}})

r = requests.post(
	url=runs_url,
	headers=HEADERS,
	data=protocol_id_payload
	)

r_dict = json.loads(r.text)
run_id = r_dict["data"]["id"]

print(f"Run ID:\n{run_id}")
```

```python
protocols_url = f"http://{ROBOT_IP}:31950/protocols"

r = requests.get(
	url=protocols_url,
	headers=HEADERS
	)

protocols_dict = json.loads(r.text)
print("Protocols list:")
for p in protocols_dict["data"]:
	print(p)
```
